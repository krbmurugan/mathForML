```python
x= 25
print(x)
print(type(x))
```

```python
y = 31
print(y)
print(type(y))
py_sum = x +y


print(py_sum)
print(type(py_sum))

x_float = 25.1
float_sum = x_float + y

print(float_sum)
print(type(float_sum))
```

#### Scalars in PyTorch
PyTorch and TensorFlow are the two most popular automatic differentiation libraries (a focus of the Calculus I and Calculus II subjects in the ML Foundations series) in Python, itself the most popular programming language in ML.
PyTorch tensors are designed to be pythonic, i.e., to feel and behave like NumPy arrays.
The advantage of PyTorch tensors relative to NumPy arrays is that they easily be used for operations on GPU (see here for example).
Documentation on PyTorch tensors, including available data types, is here.

```python
import torch
```

```python
x_pt = torch.tensor(25)
print('x_pt:', x_pt)
print('x_pt shape ', x_pt.shape)
print('x_pt type', type(x_pt)) 

x_pt_dtype = x_pt.dtype
print('x_pt dtype', x_pt_dtype)

y_pt = torch.tensor([21,35,34])
print('y_pt:', y_pt)
print('y_pt shape', y_pt.shape)
print('y_pt dtype', y_pt.dtype)
print('y_pt type', type(y_pt)) 

#how to get the dimension of a tensor
print('y_pt ndim', y_pt.ndim)


z_pt = torch.tensor([[21,35],[23,35],[12,34],[23,45]]) 
print('z_pt:', z_pt)
print('z_pt shape', z_pt.shape)
print('z_pt dtype', z_pt.dtype)
print('z_pt type', type(z_pt)) 
print('z_pt ndim', z_pt.ndim)

```

#### Scalars in TensorFlow

```python
import tensorflow as tf
```

```python
x_tf = tf.Variable(25, dtype = tf.int16)
print(x_tf)

print(x_tf.get_shape())

y_tf = tf.Variable(34, dtype=tf.int16)
print('y_tf:', y_tf)

tf_sum = x_tf + y_tf
print('tf_sum:', tf_sum)
print('tf_sum type: ', type(tf_sum))

tf_sum2 = tf.add(x_tf, y_tf)
print('tf_sum2:', tf_sum2)
print('tf_sum2 type: ', type(tf_sum2))
```
