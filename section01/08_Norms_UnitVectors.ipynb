### L2 Norm 

- Its the distance of the vector from the origin
- To calculate the distance of the vector from the Origin, the magnitude of the vector below is the formula.

$$||x||_2 = \sqrt{\sum_{n}^{}{{x_i}^2}}$$




```python
import numpy as np

x = np.array([25, 2, 5])
print('vector', x)

x_l2 = (25**2 + 2**2 + 5**2) **(1/2)
print('x_l2', x_l2)

#Numpy library to calculate the L2 Norm of any vector
x_l2_02 = np.linalg.norm(x)
print('x_l_02', x_l2_02)


y = np.array([10,5,12])
print('Norm of y ', np.linalg.norm(y))
```

### L1 Norm
- Sum of absolute values of elements of the vectors
$$||x||_1 = \sum_{n}^{}{x_i}$$
