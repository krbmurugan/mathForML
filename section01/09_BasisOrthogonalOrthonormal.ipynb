- Basis vectors are set of vectors that can be scaled to any vector.
- In a 2 dimensional space we got 2 basis vectors i that is in (1, 0) and j that is in (0,1)
- So a vector (3, 2) in 2 dimension is 3*i + 2*j

- Orthogonal vectors

```python
import numpy as np
```

```python
i = np.array([1,0])
j = np.array([0,1])
print('i is :', i, '\nj is :', j)

dotprod = np.dot(i, j)
print('dotprod:',dotprod)
```
