#### Vectors (Rank 1 or 1 Dimensional Tensor)

```python
import numpy as np
import torch
import tensorflow as tf
```

```python
x = np.array([25, 2, 5])
print('x :', x)

y = np.array([35, 4.2, 16.3], dtype=np.float16)
print('y :', y)
```

```python
x_t = x.T
print('x transposed:', x_t)

print('x shape :', x.shape)
print('x transposition :', x_t.shape)
```

```python
y = np.array([[24, 3, 5],[25, 23, 15]])
print('shape of y :', y.shape)
print(y)
y_t = y.T
print('shape of y_t :', y_t.shape)
print(y_t)

```

#### Zero vectors

```python
z = np.zeros(3, dtype=np.int16)
z
```

#### Vectors in PyTorch and TensorFlow

```python
x_pt = torch.tensor([25, 2, 5])
print(x_pt)
print('x_pt shape:', x_pt.shape)


x2d_pt = torch.tensor([[25, 2, 5],[2,4,5]])
print(x2d_pt)
print('x_pt shape:', x2d_pt.shape)
```

```python
x_tf = tf.Variable([23, 4, 5,6])
x_tf
print(x_tf)

```
